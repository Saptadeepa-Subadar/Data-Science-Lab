```python
#EXTRACT ZIPFILE
import zipfile
file = zipfile.ZipFile("a.zip", "r")
file.extractall("extracted")
print("ZIP file extracted successfully")
print("\n")
```

```python
#READ A CSV FILE FROM A URL
import pandas as pd
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)
print(df)
print("\n")
```

```python
#READ A IMAGE FROM AN URL
path='image.png'
import matplotlib.pyplot as plt
import matplotlib.image as img
Image=img.imread(path)
plt.imshow(Image)
print(Image.shape)
```

```python
#DOWNLOAD A ZIP FILE
import requests
url = "https://www.kaggle.com/api/v1/datasets/download/andrewpark01/walmart-sales"
response = requests.get(url)
file = open("walmart_sales.zip", "wb")
file.write(response.content)
print("Walmart Sales ZIP file downloaded successfully")
```
