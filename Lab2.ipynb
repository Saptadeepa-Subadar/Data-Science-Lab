# Lab 2 : Write a python program for computing Mean, Median, Mode, Standard Deviation and Percentile

```python
# Program to calculate the mean of an array using numpy 
import numpy as np
ar1 = np.array([12,234,200,100,213,12,1,112,113])
mean = np.mean(ar1)
print("The original array is : ",ar1)
print("The mean of the given array is : ", mean)
```

```python
# Program to calculate the median of an array using numpy 
import numpy as np
ar2 = np.array([12,13,14,15,16,17,18,19,20])
med = np.median(ar1)
print("The original array is : ",ar2)
print("The mean of the given array is : ", med)
```

```python
# Program to calculate the mode
import statistics as st
import numpy as np
ar3 = np.array([10,20,20,30,40,50,60,10,10,100])
md = st.mode(ar1)
print("The original array is : ",ar3)
print("\n")
print("The mean of the given array is : ", md)
```

```python
# Program to calculate the standard deviation of an array
import numpy as np
ar4 = np.array([10,20,20,15,17,18,20,17,15])
m1 = np.mean(ar4)
sd = np.std(ar4)
print("The original array is : ",ar4)
print("\n")
print("The mean of the given array is : ", m1)
print("The calculated value of S.D i : ",sd)
```

```python
# Programto calculate the percentile 
per = np.array([10,14,15,17,45,50,55,60,80,100,33,63,72])
# Find out the 75 percentile of the data 
a = np.percentile(per,75)
print(a)
```

## CLASSWORK : 

### Take an input from the user and perform the following task : 
### step 1 : Ask the user to input the size of the array 
### step 2 : Ask the user to input the values in the array
### step 3 : Calculate the : Mean, Median, Mode, S.D, Percentile
### step 4 : Find the dimention of the array given by the user 

```python
import numpy as np
import statistics as st

#Input size
n=int(input("Enter the array size: "))
arr = []

#Input the values of array
print("Enter element: ")
for i in range (n):
    v=int(input("Enter value:"))
    arr.append(v)

print("\n")
print("Array: ")
print(arr)
print("\n")

#calculate the Mean, Median, Mode, S.D, Percentile
mean = np.mean(arr)
med = np.median(arr)
md = st.mode(arr)
sd = np.std(arr)
a = np.percentile(arr, 75)
b = np.ndim(arr)

print("The mean of the given array is:",mean)
print("The median of the given array is:",med)
print("The mode of the given array above is",md)
print("The calculated value of S.D is:",sd)
print("The 75 percentile of the data:",a)

#Dimension of array

print("The dimension of the array is ",b)
```
