```python
# Library to generate random data points
from numpy import random
```

```python
import matplotlib as plt
```

```python
# go to matplotlib find pyplot and import it as plt
import matplotlib.pyplot as plt
```

#### Normal / gaussian distribution

```python
x = random.normal(size=1000)
print(x)
```

```python
y = random.normal(size = (2,3))
print(y)
```

```python
# generate a ramdom normal distribution
# size = 2*3
# mean = 1
# s.d = 2
# x = random.normal(mean(loc) = 1, SD(scale) = 2, size = (2,3))
x = random.normal(loc = 1, scale = 2, size = (2,3))
print(x)
```

```python
import seaborn as sns
sns.displot(random.normal(size = 1000))
plt.show()
```

```python
# kde - kernal density estimation
sns.displot(random.normal(size = 1000),kind = "kde")
plt.show()
```

#### BINOMIAL DISTRIBUTION

```python
from numpy import random
# n = tossing the coin for 10 no. of time = 1 event, p = probability and size = no. of data points 
x = random.binomial( n = 10, p = 0.5, size = 1000)
print(x)
```

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
sns.displot(random.binomial(n=10,p=0.5,size=1000), kde = False)
```

#### POISSON DISTRIBUTION

```python
# It is a discrete distibution
# lam = rate of known occurances 
# size = the size of the returned array 
```

```python
from numpy import random
x = random.poisson(lam=2,size=7)
print(x)
```

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
sns.displot(random.poisson(lam=2,size=1000),kde = False)
plt.show()
```

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
sns.displot(random.poisson(lam=2,size=1000),kde = True)
plt.show()
```

#### UNIFORM DISTRIBUTION

```python
x = random.uniform(size = (2,3))
print(x)
```

```python
sns.displot(random.uniform(size = 1000))
plt.show()
```

```python
sns.displot(random.uniform(size = 1000), kind = "kde")
plt.show()
```

```python
sns.displot(random.uniform(size = 1000), kde = True ) 
plt.show()
```

#### MULTINOMIAL DISTRIBUTION

```python
# n = no. of possible outcome 
# pvals = list of probabilities outcome = [1/6,1/6,1/6,1/6,1/6,1/6]
x = random.multinomial(n = 20, pvals = [1/6,1/6,1/6,1/6,1/6,1/6])
# size = shape of he returned array
x = random.multinomial(20,[1/6]*6,size = 1)
print(x)
```

```python
# dice landed 4 and 2 twice 
```

```python
sns.displot(random.multinomial(n = 20, pvals = [1/6,1/6,1/6,1/6,1/6,1/6]))
plt.show()
```

#### EXPONENTIAL DISTRIBUTION

```python
x = random.exponential(scale = 2, size = (2,3))
print(x)
```

```python
sns.displot(random.exponential(scale = 2, size = (2,3)),kde = True )
plt.show()
```

```python
sns.displot(random.exponential(size = 1000),kde = True)
plt.show()
```

#### RAYLEIGH DISTRIBUTION

```python
x = random.rayleigh(size = [2,3])
print(x)
```

```python
sns.displot(random.rayleigh(size = 1000),kde = True)
plt.show()
```
