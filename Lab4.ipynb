# FILE HANDLING 

```python
#OPENN(file name , mode)
#mode :-
    # r : opne an existing file for read operation
    # w : write operation
    # a : append(Make some changes to existing file)
    # r+ : read and write (override the existing file)
    # w+ : write and read (override the existing file)
    # a+ : append and read (NO override)
    # x : create a specific file 
    # "t" : text file 
    # "b" : Binary file 
```

```python
f = open("samplefile.txt","r")
print(f.read(3)) # --> 3 = 3 character and space is also counted as char
                 # read is used to read a character 
f.close()
```

```python
#Read two line in a file 
f = open("samplefile.txt","r")
print(f.readline())   # readline() is used to read a line 
print(f.readline())
```

```python
# Read the file line by line 
f = open("samplefile.txt","r")
for x in f:
    print(x)
f.close()
```

```python
# Create a file write some content on it 
file = open("a2.txt","w")
file.write("This is the first line that i have written\n")
file.write("This is the second line")
file.close()
```

```python
# add more line in the text file or append new line to a file
file = open("a2.txt","a")
file.write("\nThis is new line")
file.close()
```

```python
# DELETE A FILE FROM THE SYSTEM 
import os 
if os.path.exists("a2.txt") :
    os.remove("a2.txt")
    print("File deleted")
else : 
    print("The file doesn't exist")
```

```python
# Python program to illustrate the split() function
# open the file (samplefile.txt) and store it on "pathh'
pathh = "samplefile.txt"
# opening the file(path) and name it as file 
with open(pathh, "r") as file:
    #in data variable store all the lines of the file 
    data = file.readlines()
    #data(all the lines)---store--->line(variable)
    for line in data:
        #in variable word take the line(all the data) and split it 
        word = line.split()
        #printing the lines 
        print(word)
```

```python
df = pd.DataFrame({'name' : ['ajay', 'vijay', 'sujal'],
                   'course' : ['eng', 'bio', 'bsc'],
                   'location' : ['delhi', 'lucknow', 'arunachal']
                  })
# telling what method to use and which file is used 
compression_opts = dict(method = 'zip',archive_name = 'a.csv')
# specifying the zip file (a.zip), compression = whatever given above
df.to_csv(r'a.zip', index = False,compression = compression_opts)
```
