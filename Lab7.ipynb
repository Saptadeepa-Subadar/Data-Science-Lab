```python
import pandas as pd 
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt 
# sklearn is a python library for machine learning 
from sklearn.model_selection import train_test_split # ! pip install scikit-learn
```

```python
url = 'http://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
```

```python
df = pd.read_csv(url)
print(df)
attributes = ["sepal_length","sepal_width","petal_length","petal_width","class"]
dataset = pd.read_csv(url, names = attributes)
dataset.column = attributes 
dataset
```

```python
iris = pd.read_csv('iris.csv')
iris.head(5)
```

```python
# TYPES FOR EACH COLUMN
iris.dtypes
# CHECKING THE INFORMATION OF THE COLUMN 
iris.info()
```

```python
# Numerical summary --> numrical column
iris.describe()
```

```python
# number of instamces in each class 
iris.groupby('species').size()
```

```python
# minimum values
iris.groupby('species').min()
```

```python
# maximum values
iris.groupby('species').max()
```

```python
# check for null
iris.isnull().any()
```

```python
# PLOT - HISTOGRAM (UNIVARIATE)
iris['petal_length'].hist(bins = 20)
# histogram plot using matplotlib
plt.hist(data = dataset, x = 'petal_length',bins = 20)
# Using seaborn
sns.displot(dataset['petal_length'],bins = 20)
# Histogram (UD) of each variable (columns) usind pandas 
iris.hist(bins = 20,figsize= (10,10))
```

```python
# PLOT - HISTOGRAM (UNIVARIATE)
iris['petal_length'].hist(bins = 20)
# histogram plot using matplotlib
plt.hist(data = dataset, x = 'petal_length',bins = 20)
# Using seaborn
sns.displot(dataset['petal_length'],bins = 20)
# Histogram (UD) of each variable (columns) usind pandas 
iris.hist(bins = 20,figsize= (10,10))
# modify
iris.groupby('species').petal_length.plot.hist(alpha=0.4)
plt.xlabel('Petal Length (cm)')
plt.suptitle("Histogram of petal length (cm) for different species ")
plt.legend()
plt.grid()
```

```python
# BOX PLOT 
sns.boxplot(data=iris,y = 'sepal_length')
min(iris['sepal_length'])
np.median(iris['sepal_length'])
```

```python
# BOX PLOT 
sns.boxplot(data=iris,y = 'sepal_length')
min(iris['sepal_length'])
np.median(iris['sepal_length'])
# First (Q1), Second (Q2), Third (Q3)
iris.sepal_length.quantile([0.25,0.5,0.75])
np.percentile(iris.sepal_length,[25,50,77])
# Box plot for each categories
# use --> seaborn --> use --> boxplot --> take data = iris -- give me the sepal length of differenet species 
sns.boxplot(data = iris, x = 'species', y = 'sepal_length')
```

```python
# Violin Plot 
sns.violinplot(data = iris, x = 'species', y = 'sepal_length')
```

```python
# BIVARIATE RELATIONSHIP
# Scatter plot 
# To see the relationship between two variable (bi-variate)
sns.set_style('darkgrid')
# hue = species --> scatter plot in graph will be colored as per the species
sns.scatterplot(data = iris, x = 'petal_length', y = 'sepal_length',hue = 'species')
```

```python
# Relation between petal_length and sepal_width
sns.set_style('darkgrid')
sns.scatterplot(data = iris, x = 'petal_length', y = 'sepal_width',hue = 'species')
```

```python
# Scatter plot using matplotlib, pandas and seaborn
# SEABORN
sns.scatterplot(data = iris, x = 'petal_length', y = 'sepal_length',hue = 'species')
# MATPLOTLIB
plt.scatter(data = iris, x = 'petal_length', y = 'petal_width',c = 'green', marker = '+')
# PANDAS
iris.plot.scatter(x = 'sepal_length', y = 'sepal_width',c = 'red', marker = '+',s = 15)
```

```python
# PAIR PLOT 
sns.pairplot(iris,hue = "species", height = 2, palette ='colorblind')
```
