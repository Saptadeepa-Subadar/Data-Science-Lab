```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns 
import sklearn 
from sklearn.cluster import KMeans
import math
```

```python
sns.set()
```

```python
# Load the data 
data = pd.read_csv('3.01.Countryclusters.csv')
data
```

```python
# Plotting the dataset 
plt.scatter(data['Latitude'],data['Longitude'])
plt.xlim(-180,180)
plt.ylim(-90,90)
plt.show()
```

```python
# SELECTING THE FEATURE
x = data.iloc[:,1:3]
x
```

```python
# Clustering 
kmeans = KMeans(5)
# fitting the data present in x in kmean
kmeans.fit(x)
# Clustering Result
# fitting the data into the model
identified_clusters = kmeans.fit_predict(x)
identified_clusters
# creating a copy of our data 
data_with_clusters = data.copy()
# new cluster--> data
data_with_clusters['Clusters'] = identified_clusters
plt.scatter(data_with_clusters['Latitude'],data_with_clusters['Longitude'],c = data_with_clusters['Clusters'],cmap='rainbow')
plt.show()
```

```python
import math
X = [10, 12, 16, 20, 24, 26, 30, 34]
Y = [2, 4, 6, 8, 10, 12, 14, 16]
cx1 = 10
cy1 = 2
cx2 = 12
cy2 = 4
for step in range(10):
    c1_x = []
    c1_y = []
    c2_x = []
    c2_y = []
    for i in range(8):
        x = X[i]
        y = Y[i]
        distance1 = math.sqrt(math.pow(x - cx1, 2) + math.pow(y - cy1, 2))
        distance2 = math.sqrt(math.pow(x - cx2, 2) + math.pow(y - cy2, 2))
        if distance1 <= distance2:
            c1_x.append(x)
            c1_y.append(y)
        else:
            c2_x.append(x)
            c2_y.append(y)
    cx1 = sum(c1_x) / len(c1_x)
    cy1 = sum(c1_y) / len(c1_y)
    cx2 = sum(c2_x) / len(c2_x)
    cy2 = sum(c2_y) / len(c2_y)
print("Cluster 1 Center:", cx1, ",", cy1)
print("Cluster 1 Points:")
for i in range(len(c1_x)):
    print("(", c1_x[i], ",", c1_y[i], ")")
print("\nCluster 2 Center:", cx2, ",", cy2)
print("Cluster 2 Points:")
for i in range(len(c2_x)):
    print("(", c2_x[i], ",", c2_y[i], ")")
```
