```python
import pandas as pd
df = pd.read_csv("dirtydata.csv")
print(df)
```

```python
# To get information
df.info()
```

```python
# Describe the data
df.describe()
```

```python
# Prints the first ten rows of the data set
print(df.head(10))
print("n")
# Prints the last ten rows of the data set
print(df.tail(10))
```

```python
# Convert the dataframe into string
print(df.to_string())
```

```python
# Drop NAN(Not A Number) value 
# new_df(new variable) --> stores the values of original data frame and removes nan values  
new_df = df.dropna()
print(new_df)
```

```python
# To change the original df
# Use 'inplace=True'
df.dropna(inplace=True)
print(df)
```

```python
df.loc[7,'Duration'] = 45
print(df)
```

```python
# for a larger data set --> where loc of the wrong entry is hard to find
# take index(header) --> store --> x
# increment the value of x
# Till the condition becomes false
for x in df.index:
    # if in the location (x: depends on loop)
    # and duration is > 120
    # in the location x 
    # set he value as 120
    if df.loc[x,"Duration"]>20:
        df.loc[x,"Duration"] = 120
```

```python
# Delete the value greater than 120
for x in df.index:
    if df.loc[x,"Duration"] > 120:
        df.drop(x,inplace = True)
```

```python
# Drop the duplicate row 
nv = df.duplicated()
```

```python
print(nv)
```

```python
# Drop duplicate row
nv = df.duplicated()
new_out = df.loc[nv == True]
print(new_out)
df.drop_duplicates(inplace = True)
print(df.to_string)
```

```python
import pandas as pd 
import numpy as np
# Dictionary
dict = {'First Score' : [100,90,np.nan,95],
        'Second score' : [np.nan, 45,56,np.nan],
        'Third Score' : [np.nan, 40,80,98]
       }
df = pd.DataFrame(dict)
print(df)
```

```python
print(df.to_string)
# Return true if null and false if not null
print(df.isnull())
# Return true if not null and false if null
print(df.notnull())
```

```python
# Fill the value with a specific value 
print(df.fillna(0))
```

```python
# Fill the NaN value with mean
df_mean = df.mean()
print(df_mean)
data_new = df.fillna(df.mean())
print(data_new)
```
