### CLASSWORK : 
##### 1. Take the search element from the user 
##### 2. Check if the element is present in array, asearch = [2,5,6.9,10]

```python
import numpy as np
asearch = np.array([2,5,6.9,10])
abc = int(input("The element is present in asearch :"))
res = np.where(asearch == abc)

#element present or not 
if abc in asearch : 
    print("The element is present in asearch ")
    print("Element",abc, "found at position",res)
else : 
    print("Element not found")
```
